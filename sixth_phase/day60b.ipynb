```python
def plot_decision_boundary(x_array,y_array,degree=1):
    
    poly = PolynomialFeatures(degree=degree)
    X_trf = poly.fit_transform(x_array)
    
    clf = LogisticRegression()
    clf.fit(X_trf,y)
    
    accuracy = np.mean(cross_val_score(clf,X_trf,y,scoring='accuracy',cv=10))
    
    a=np.arange(start=x_array[:,0].min()-1, stop=x_array[:,0].max()+1, step=0.01)
    b=np.arange(start=x_array[:,1].min()-1, stop=x_array[:,1].max()+1, step=0.01)


    XX,YY=np.meshgrid(a,b)
    
    input_array=np.array([XX.ravel(),YY.ravel()]).T

    labels=clf.predict(poly.transform(input_array))
    
    plt.contourf(XX,YY,labels.reshape(XX.shape),alpha=0.5)
    plt.scatter(x_array[:,0],x_array[:,1], c=y)
    plt.title('Degree = {}, accuracy is {}'.format(degree,np.round(accuracy,4)))
```

```python
plot_decision_boundary(x_array,y_array)
```

```python
plot_decision_boundary(x_array,y_array,degree=2)
```

```python
plot_decision_boundary(x_array,y_array,degree=3)
```

```python
plot_decision_boundary(x_array,y_array,degree=4)
```
