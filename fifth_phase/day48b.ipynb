```python
# defing class
class meralr:
    def __init__(self):
        self.m = None
        self.b = None

    # fit function
    def fit(self,x_train,y_train): # self is used for accessing the variable of another method
        # it is resposible for finding m and b
        num = 0
        den = 0

        for i in range(x_train.shape[0]):
            num  = num  + ( (x_train.iloc[i] - x_train.mean())* (y_train.iloc[i] - y_train.mean()) )
            den = den + ((x_train.iloc[i] - x_train.mean() )*(x_train.iloc[i] - x_train.mean()))

        self.m = num/den
        self.b = y_train.mean() - (self.m*x_train.mean())
        print(self.m)
        print(self.b)

    # predict function
    def predict(self,x_test):
        print(x_test)
        return self.m*x_test + self.b
        
```
