# Linear Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('../../datasets/placement2.csv')
```

```python
data.head()
```

```python
plt.scatter(data['cgpa'],data['package'])
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
x = data[['cgpa']]
y = data['package']
```

```python
x.shape
```

```python
x.head()
```

```python
y.head()
```

```python
data.isnull().sum()
```

```python
data.value_counts()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=42,test_size=0.3)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train) # Always remember fit required 2D data always 
# ir read data like as x[rows][columns]
```

```python
# drawing best fit line
plt.scatter(data['cgpa'],data['package'])
plt.plot(x_train,lr.predict(x_train),color='red')
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
# slope of best fit line
slope = lr.coef_
```
